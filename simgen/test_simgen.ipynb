# Exemple d'utilisation locale de SimGen

Ce notebook est utilisé pour tester les fonctionnalités de simulation du modèle simgen.

Il importe les librairies nécessaires en mode local et définit les paramètres de simulation.

```python
# Importer les librairies nécessaires
import warnings
import pandas as pd 
import numpy as np
from matplotlib import pyplot as plt
warnings.filterwarnings("ignore")
```

```python
# Définir le chemin du fichier local
import sys
sys.path.append('../../simgen-creei-1.0.0')
```

```python
# Importer les librairies de simgen
import simgen
from simgen import model, formating
```

```python
# Définir le chemin du fichier local de données
donnees_brutes = '../../../bdsps2017_slice.csv'
```

```python
# Paramètres de la simulation
annee_fin = 2025
nb_rep = 1
fecondite = 'reference'
mortalite = 'low'
taux_immigration = 0.0066
```

```python
# Préparation des données
preparation_data=formating()
preparation_data.bdsps_format(donnees_brutes)
```

```python
# Création d'une instance de la classe model
base = model(stop_yr=(annee_fin))
base.startpop('start_pop')
base.birth_assumptions(scenario=fecondite)
base.dead_assumptions(scenario=mortalite)
base.immig_assumptions(init='imm_pop', num=taux_immigration)
```

```python
# Lancer la simulation
base.simulate(rep=nb_rep)
```

```python
# Exemple de tableau de données: population homme par années
population_hommes=base.stats.freq(sub='male==True')
population_hommes.head()
```

```python
# Exemple de statistiques: proportion de la population par niveau de scolarité
proportion_niveau_scolarite = base.stats.prop('educ', sub="age>=25 and age<=64 and insch==False")
proportion_niveau_scolarite.head()
```
